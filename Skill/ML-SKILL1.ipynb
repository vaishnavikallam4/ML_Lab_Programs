```python
# ============================================================
# EXPERIMENT: Placement Prediction Dataset Generation + EDA
# ============================================================

import numpy as np
import pandas as pd


# ============================================================
# 1. CREATE RANDOM NUMBER GENERATOR
# ============================================================

rng = np.random.default_rng(42)


# ============================================================
# 2. NUMBER OF STUDENT RECORDS
# ============================================================

N = 800


# ============================================================
# 3. CREATE DATASET
# ============================================================

df = pd.DataFrame({

    # CGPA between 4 and 10
    "cgpa":
        np.clip(
            rng.normal(7.4, 0.9, N),
            4,
            10
        ).round(2),

    # Attendance percentage between 45 and 100
    "attendance":
        np.clip(
            rng.normal(80, 9, N),
            45,
            100
        ).round(1),

    # Number of backlogs
    "backlogs":
        rng.poisson(0.6, N),

    # Number of internships: 0 to 3
    "internships":
        rng.integers(0, 4, N),

    # Coding score between 0 and 100
    "coding_score":
        np.clip(
            rng.normal(58, 16, N),
            0,
            100
        ).round(1),

    # Communication score between 0 and 100
    "comm_score":
        np.clip(
            rng.normal(60, 14, N),
            0,
            100
        ).round(1),

    # Student branch
    "branch":
        rng.choice(
            ["CSE", "AI&DS", "ECE", "IT"],
            N
        )
})


# ============================================================
# 4. CREATE HIDDEN PLACEMENT SCORE
# ============================================================

z = (
    -8
    + 0.6 * df["cgpa"]
    + 0.03 * df["attendance"]
    + 0.04 * df["coding_score"]
    - 0.8 * df["backlogs"]
)


# ============================================================
# 5. CONVERT SCORE INTO PLACEMENT PROBABILITY
# ============================================================

probability = 1 / (1 + np.exp(-z))


# ============================================================
# 6. CREATE BINARY TARGET
# ============================================================

df["placed"] = (
    rng.random(N) < probability
).astype(int)


# ============================================================
# 7. ADD MISSING VALUES
# ============================================================
# Randomly select 20 students and make communication score missing

missing_indices = rng.choice(
    N,
    20,
    replace=False
)

df.loc[
    missing_indices,
    "comm_score"
] = np.nan


# ============================================================
# 8. SAVE RAW DATASET
# ============================================================

df.to_csv(
    "students_raw.csv",
    index=False
)


# ============================================================
# 9. DISPLAY FIRST 5 ROWS
# ============================================================

print("============================================================")
print("1. FIRST 5 ROWS")
print("============================================================")

print(df.head())


# ============================================================
# 10. DATASET INFORMATION
# ============================================================

print("\n============================================================")
print("2. DATASET INFORMATION")
print("============================================================")

df.info()


# ============================================================
# 11. DESCRIPTIVE STATISTICS
# ============================================================

print("\n============================================================")
print("3. DESCRIPTIVE STATISTICS")
print("============================================================")

print(
    df.describe().round(2)
)


# ============================================================
# 12. MISSING VALUES
# ============================================================

print("\n============================================================")
print("4. MISSING VALUES")
print("============================================================")

print(
    df.isna().sum()
)


# ============================================================
# 13. PLACEMENT RATE
# ============================================================

print("\n============================================================")
print("5. PLACEMENT RATE")
print("============================================================")

print(
    "Placed rate:",
    round(df["placed"].mean(), 3)
)


# ============================================================
# 14. TARGET DISTRIBUTION
# ============================================================

print("\n============================================================")
print("PLACEMENT DISTRIBUTION")
print("============================================================")

print(
    df["placed"].value_counts()
)

print("\n0 = Not Placed")
print("1 = Placed")


# ============================================================
# 15. MACHINE LEARNING LIFECYCLE
# ============================================================

lifecycle = [

    (
        "1 Problem framing",
        "binary classification: will a student be placed?"
    ),

    (
        "2 Data collection",
        "students_raw.csv generated above"
    ),

    (
        "3 EDA",
        "this session"
    ),

    (
        "4 Preprocessing",
        "Session 2"
    ),

    (
        "5 Feature engineering",
        "Session 3"
    ),

    (
        "6 Modelling",
        "Sessions 4-12"
    ),

    (
        "7 Evaluation",
        "Sessions 13-21"
    ),

    (
        "8 Deployment & monitoring",
        "Sessions 22-24"
    )
]


print("\n============================================================")
print("MACHINE LEARNING LIFECYCLE")
print("============================================================")

for stage, description in lifecycle:

    print(
        f"{stage:<28} -> {description}"
    )


# ============================================================
# 16. FINAL DATASET SUMMARY
# ============================================================

print("\n============================================================")
print("FINAL DATASET SUMMARY")
print("============================================================")

print("Number of records :", len(df))
print("Number of features:", len(df.columns) - 1)
print("Target column     : placed")
print("Saved file        : students_raw.csv")


# ============================================================
# EXPERIMENT COMPLETED
# ============================================================

print("\n============================================================")
print("EXPERIMENT COMPLETED SUCCESSFULLY!")
print("============================================================")
```
